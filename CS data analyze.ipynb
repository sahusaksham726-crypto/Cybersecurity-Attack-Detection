```python
for col in df.columns:
    print(f"{col}: {df[col].nunique()} unique values")
```

```python
df["url_missing"] = df["url"].isna()
print(df.groupby("url_missing")["label"].value_counts(normalize=True) * 100)
```

```python
pd.crosstab(df["url_missing"], df["attack_type"])
```

```python
pd.crosstab(df["url_missing"],df["protocol"],normalize="index") * 100
```

```python
pd.crosstab(df["url_missing"],df["is_internal_traffic"],normalize="index") * 100
```

```python
url_port_analysis = (df.groupby("dst_port")["url_missing"].agg(["count", "sum", "mean"])
      .sort_values("mean", ascending=False))
url_port_analysis.head(15)
```

```python
url_src_port_analysis = (df.groupby("src_port")["url_missing"].agg(["count", "sum", "mean"])
      .sort_values("mean", ascending=False))
url_src_port_analysis.head(15)
```

```python
pd.crosstab(df["url_missing"],df["user_agent"])
```

```python
df["timestamp"] = pd.to_datetime(df["timestamp"])
```

```python
url_time = (df.groupby(df["timestamp"].dt.date)["url_missing"].agg(["count", "sum", "mean"]))
url_time.head()
```
