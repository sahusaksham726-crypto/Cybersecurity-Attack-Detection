```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
results = pd.DataFrame({
    "Model": ["Model A", "Model B", "Model C"],
    
    "Rows": [
        10000,
        6768,
        3232
    ],
    
    "URL": [
        "Ignored",
        "Available + Features",
        "Missing"
    ],
    
    "Accuracy": [
        71.75,
        74.37,
        75.12
    ],
    
    "Precision": [
        9.52,
        9.19,
        10.06
    ],
    
    "Recall": [
        71.25,
        61.11,
        65.38
    ],
    
    "F1_Score": [
        16.79,
        15.98,
        17.44
    ],
    
    "ROC_AUC": [
        81.18,
        77.38,
        79.50
    ]
})

results
```

```python
metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1_Score",
    "ROC_AUC"
]

plot_data = results.set_index("Model")[metrics]

plt.figure(figsize=(12, 6))

plot_data.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Overall Model Performance Comparison")
plt.xlabel("Model")
plt.ylabel("Score (%)")
plt.xticks(rotation=0)
plt.legend(title="Metrics")

plt.tight_layout()

plt.savefig(
    "PNG FOLDER/final_overall_model_comparison.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
print("=" * 60)
print("BEST MODEL FOR EACH METRIC")
print("=" * 60)

for metric in metrics:
    
    best_index = results[metric].idxmax()
    
    best_model = results.loc[
        best_index,
        "Model"
    ]
    
    best_score = results.loc[
        best_index,
        metric
    ]
    
    print(
        f"{metric:12} → "
        f"{best_model} ({best_score:.2f}%)"
    )
```

```python
ranking = results[
    [
        "Model",
        "Recall",
        "Precision",
        "F1_Score",
        "ROC_AUC",
        "Accuracy"
    ]
].copy()

ranking["Overall_Rank_Score"] = (
    ranking["Recall"] * 0.35 +
    ranking["F1_Score"] * 0.25 +
    ranking["ROC_AUC"] * 0.30 +
    ranking["Precision"] * 0.10
)

ranking = ranking.sort_values(
    "Overall_Rank_Score",
    ascending=False
)

ranking
```

```python
print("=" * 60)
print("FINAL MODEL RANKING")
print("=" * 60)

for rank, (_, row) in enumerate(
    ranking.iterrows(),
    start=1
):
    
    print(
        f"{rank}. {row['Model']} "
        f"→ Score: {row['Overall_Rank_Score']:.2f}"
    )
```

```python
print("=" * 75)
print("                    FINAL MODEL DECISION")
print("=" * 75)

print("""
Model A is selected as the final model.

Reason:

1. Model A uses the complete dataset of 10,000 records.
2. Model A achieves the highest Recall (71.25%).
3. Model A achieves the highest ROC-AUC (81.18%).
4. Recall is highly important in cybersecurity because
   missing an actual attack can be costly.
5. Model B did not improve the important detection metrics
   after adding URL-derived features.
6. Model C achieved the highest Accuracy, Precision and F1 Score,
   but it was trained only on the 3,232 records where URL was missing.
7. Model A therefore provides the strongest overall baseline
   for detecting attacks across the complete dataset.

Final Selected Model: MODEL A
""")
```
